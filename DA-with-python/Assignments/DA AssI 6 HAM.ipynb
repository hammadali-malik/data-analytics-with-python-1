```python
import numpy as np

arr = np.arange(10, 51, 5)
print(arr)
print("Shape:", arr.shape)
print("Dtype:", arr.dtype)
print("Dimensions:", arr.ndim)
```

```python
import numpy as np

zeros_arr = np.zeros((3,4))
print("Zeros:")
print(zeros_arr)

ones_arr = np.ones((3,4))
print("\nOnes:")
print(ones_arr)

identity = np.eye(4)
print("\nIdentity:")
print(identity)

linspace_arr = np.linspace(0, 1, 8)
print("\nLinspace:")
print(linspace_arr)
```

```python
import numpy as np

arr1 = np.arange(1, 25) # 1 se 24 tak
print("1D ka shape:", arr1.shape)
print(arr1)

arr2 = arr1.reshape(4, 6) # ab 4 line 6 column
print("\n2D (4x6):")
print(arr2)

arr3 = arr1.reshape(2, 3, 4) # 2 dabbe, har me 3x4
print("\n3D (2x3x4):")
print(arr3)
```

```python
import numpy as np

arr = np.array([12, 45, 7, 23, 56, 89, 34, 21])
print("Pura array:", arr)

print("\nFirst 3:", arr[:3])
print("Last 3:", arr[-3:])
print("Ek chor ke ek:", arr[::2])
print("Ulta:", arr[::-1])
```

```python
import numpy as np

arr = np.arange(1, 26).reshape(5, 5)
print(arr)

print("\nSecond Row (2nd line):", arr[1])
print("Third Column (3rd khamba):", arr[:, 2])
print("\nCenter ka 3x3 hissa:")
print(arr[1:4, 1:4])
```

```python
import numpy as np

a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a ka square =", a ** 2)
print("Dot product =", np.dot(a, b))
```

```python
import numpy as np

arr = np.array([[4, 8, 2],
                [9, 1, 7],
                [5, 6, 3]])

print(arr)
print("\nSum of all:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Min:", np.min(arr))
print("Max:", np.max(arr))
print("Std Dev:", np.std(arr))

print("\nSum Row-wise (axis=1):", np.sum(arr, axis=1))
print("Sum Column-wise (axis=0):", np.sum(arr, axis=0))
```

```python
import numpy as np

arr = np.random.randint(1, 101, 20) 
print("Array:", arr)

print("\nGreater than 50:", arr[arr > 50])
print("Even numbers:", arr[arr % 2 == 0])

count_div_5 = np.sum(arr % 5 == 0)
print("Count divisible by 5:", count_div_5)
```

```python
import numpy as np

arr = np.array([5, -3, 8, -1, 0, -7, 4])
print("Original:", arr)

new_arr = np.where(arr < 0, 0, arr)
print("After replacing negative with 0:", new_arr)

print("\nIndex of Max value:", np.argmax(arr))
print("Index of Min value:", np.argmin(arr))
```

```python
import numpy as np

a = np.array([[1,2,3],[4,5,6]])
b = np.array([[7,8,9],[10,11,12]])

print("A:\n", a)
print("\nB:\n", b)

print("\nVertical Join (vstack):\n", np.vstack((a,b)))
print("\nHorizontal Join (hstack):\n", np.hstack((a,b)))

arr = np.array([[9, 3, 6], [2, 8, 1]])
print("\nOriginal for sorting:\n", arr)

sorted_arr = np.sort(arr, axis=1) # sort each row
print("\nSorted rows:\n", sorted_arr)

print("\nFlatten to 1D:", sorted_arr.flatten())
```
