```python
lst = list(map(int, input("Enter List : ").split()))
print("Alternate elements:", lst[::2])
```

```python
lst = list(map(int, input("Enter List: ").split()))
ulta = lst[::-1]
print("Reversed:", ulta)
```

```python
lst = list(map(int, input("Enter List: ").split()))
bara = lst[0]
for num in lst:
    if num > bara:
        bara = num
print("Largest:", bara)
```

```python
lst = list(map(int, input("Enter List: ").split()))
if len(lst) > 0:
    new_list = [lst[-1]] + lst[:-1]
    print("Rotated:", new_list)
```

```python
sentence = input("Enter a Sentence: ")
word = input("Enter a word to delete: ")
result = sentence.replace(word, "")
print("Result:", result)
```
