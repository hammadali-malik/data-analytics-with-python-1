```python
r = int(input("Enter rows: "))
c = int(input("Enter cols: "))
print("Enter Matrix 1:")
m1 = []
for i in range(r):
    row = list(map(int, input(f"Row {i+1}: ").split()))
    m1.append(row)

print("Enter Matrix 2:")
m2 = []
for i in range(r):
    row = list(map(int, input(f"Row {i+1}: ").split()))
    m2.append(row)

print("Result Matrix:")
for i in range(r):
    for j in range(c):
        print(m1[i][j] + m2[i][j], end=" ")
    print()
```

```python
r1 = int(input("Enter rows of Matrix 1: "))
c1 = int(input("Enter cols of Matrix 1: "))
r2 = int(input("Enter rows of Matrix 2: "))
c2 = int(input("Enter cols of Matrix 2: "))

if c1!= r2:
    print("Multiplication not possible! cols of 1st must equal rows of 2nd")
else:
    print("Enter Matrix 1:")
    m1 = [list(map(int, input().split())) for _ in range(r1)]

    print("Enter Matrix 2:")
    m2 = [list(map(int, input().split())) for _ in range(r2)]

    result = [[0]*c2 for _ in range(r1)]
    for i in range(r1):
        for j in range(c2):
            for k in range(c1):
                result[i][j] += m1[i][k] * m2[k][j]

    print("Result:")
    
    for row in result:
        print(row)
```
