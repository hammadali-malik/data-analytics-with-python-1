```python
import pandas as pd

data = {
    'Name': ['Ali', 'Sara', 'Ahmed', 'Ayesha', 'Bilal', 'Fatima'],
    'Age': [25, 30, 35, 28, 40, 32],
    'City': ['Karachi', 'Lahore', 'Delhi', 'Delhi', 'Karachi', 'Lahore'],
    'Salary': [40000, 50000, 45000, 60000, 55000, 38000]
}
df = pd.DataFrame(data)

df['Bonus'] = df['Salary'] * 0.10
print("After adding Bonus:")
print(df)

df['Total'] = df['Salary'] + df['Bonus']
print("\nAfter adding Total:")

print(df)

df.rename(columns={'City': 'Location'}, inplace=True)
print("\nAfter renaming City to Location:")
print(df)

df.drop(columns=['Bonus'], inplace=True)
print("\nAfter dropping Bonus:")
print(df)
```

```python
import pandas as pd
import numpy as np

data = {
    'Name': ['Ali', 'Sara', None, 'Ayesha', 'Bilal'],
    'Age': [25, None, 35, 28, 40],
    'Salary': [40000, 50000, None, 60000, 55000]
}
df = pd.DataFrame(data)
print("Original DF with NaN:")
print(df)

print("\n--- Total Missing Values ---")
print(df.isnull().sum())
print("Total NaN count:", df.isnull().sum().sum())

df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].mean(), inplace=True)
print("\nAfter filling numeric NaN with MEAN:")
print(df)

df_cleaned = df.dropna()
print("\nAfter dropping rows still missing:")
print(df_cleaned)
```

```python
import pandas as pd

data = {
    'Name': ['Ali', 'Sara', 'Ahmed', 'Ayesha', 'Bilal', 'Fatima'],
    'Age': [25, 30, 35, 28, 40, 32],
    'City': ['Karachi', 'Lahore', 'Delhi', 'Delhi', 'Karachi', 'Lahore'],
    'Salary': [40000, 50000, 45000, 60000, 55000, 38000]
}
df = pd.DataFrame(data)

print("--- A: Sorted by Salary (High to Low) ---")
print(df.sort_values(by='Salary', ascending=False))

print("\n--- B: Sorted by Age (Low to High) ---")
print(df.sort_values(by='Age', ascending=True))

print("\n--- C: Top 3 Highest Salaries ---")
print(df.nlargest(3, 'Salary'))
```

```python
import pandas as pd

data = {
    'Name': ['Ali', 'Sara', 'Ahmed', 'Ayesha', 'Bilal', 'Fatima'],
    'Age': [25, 30, 35, 28, 40, 32],
    'City': ['Karachi', 'Lahore', 'Delhi', 'Delhi', 'Karachi', 'Lahore'],
    'Salary': [40000, 50000, 45000, 60000, 55000, 38000]
}
df = pd.DataFrame(data)

print("Original DF:")
print(df)

print("\n--- A: Average Salary per City ---")
print(df.groupby('City')['Salary'].mean())

print("\n--- B: Number of employees per City ---")
print(df.groupby('City')['Name'].count())

print("\n--- C: Max Salary per City using agg ---")
print(df.groupby('City').agg({'Salary': 'max'}))

print("\n--- Bonus: Mean, Count, Max together ---")
print(df.groupby('City').agg({
    'Salary': ['mean', 'max', 'count']
}))
```

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Ali', 'Sara', 'Ahmed'],
    'City': ['Karachi', 'Lahore', 'Delhi']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Salary': [40000, 50000, 45000],
    'Department': ['IT', 'HR', 'Finance']
})

print("DF1:")
print(df1)
print("\nDF2:")
print(df2)


df3 = pd.DataFrame({
    'ID': [4, 5],
    'Name': ['Ayesha', 'Bilal'],
    'City': ['Delhi', 'Karachi']
})
print("\n--- A: Concat df1 + df3 (vertical join) ---")
print(pd.concat([df1, df3]))

print("\n--- B: Merge df1 and df2 on ID ---")
merged_df = pd.merge(df1, df2, on='ID')
print(merged_df)

print("\nAll 10 Tasks Completed! ✅")
```
