```python
numbers = [1,2,2,3,3,3,4,4,4,4]
freq = {}
for n in numbers:
    if n in freq:
        freq[n] = freq[n] + 1
    else:
        freq[n] = 1
print(freq)
```

```python
sentence = input("Enter sentence: ")
words = sentence.split()
count = {}
for w in words:
    if w in count:
        count[w] = count[w] + 1
    else:
        count[w] = 1
print(count)
```

```python
marks = {"Math": 80, "English": 70, "Physics": 90}
total = 0
for m in marks.values():
    total = total + m
avg = total / len(marks)
print("Total:", total)
print("Average:", avg)
```

```python
students = {
    "Ali": 85,
    "Sara": 92,
    "Ahmed": 78,
    "Zara": 95
}
top_name = ""
top_marks = 0
for name in students:
    if students[name] > top_marks:
        top_marks = students[name]
        top_name = name
print("Highest:", top_name)
print("Marks:", top_marks)
```

```python
contacts = {}

while True:
    print("1. Add Contact")
    print("2. View Contacts")
    print("3. Search Contact")
    print("4. Exit")

    choice = input("Enter choice: ")
    choice = choice.strip()

    if choice == "1":
        name = input("Enter name: ")
        number = input("Enter number: ")
        contacts[name] = number
        print("Contact added successfully!")

    elif choice == "2":
        print(contacts)

    elif choice == "3":
        s_name = input("Enter name to search: ")
        if s_name in contacts:
            print(s_name, ":", contacts[s_name])
        else:
            print("Contact not found!")

    elif choice == "4":
        print("Exiting program. Bye!")
        break
    else:
        print("Invalid! Please enter 1, 2, 3 or 4 only.")
```
