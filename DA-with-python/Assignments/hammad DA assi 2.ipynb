```python
units = int(input("Enter units: "))

if units <= 100:
    bill = 0
elif units <= 300:
    bill = (units - 100) * 5
else:
    bill = (200 * 5) + (units - 300) * 10

print("Total Bill:", bill)
```

```python
qty = int(input("Enter quantity: "))
cost = qty * 100

if cost > 1000:
    discount = cost * 0.10
    total = cost - discount
else:
    total = cost

print("Total cost:", total)
```

```python
marks = int(input("Enter marks: "))

if marks < 25:
    print("Grade: F")
elif marks <= 45:
    print("Grade: E")
elif marks <= 50:
    print("Grade: D")
elif marks <= 60:
    print("Grade: C")
elif marks <= 80:
    print("Grade: B")
else:
    print("Grade: A")
```

```python
a = int(input("Enter age 1: "))
b = int(input("Enter age 2: "))
c = int(input("Enter age 3: "))

oldest = max(a, b, c)
youngest = min(a, b, c)

print("Oldest:", oldest)
print("Youngest:", youngest)
```

```python
held = int(input("Classes held: "))
attended = int(input("Classes attended: "))

percent = (attended / held) * 100
print("Attendance %:", percent)

if percent >= 75:
    print("Allowed to sit in exam")
else:
    print("Not allowed to sit in exam")
```

```python
held = int(input("Classes held: "))
attended = int(input("Classes attended: "))
medical = input("Medical cause? Y/N: ")

percent = (attended / held) * 100
print("Attendance %:", percent)

if percent >= 75 or medical == 'Y':
    print("Allowed to sit in exam")
else:
    print("Not allowed to sit in exam")
```

```python
year = int(input("Enter year: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Leap Year")
else:
    print("Not Leap Year")
```
