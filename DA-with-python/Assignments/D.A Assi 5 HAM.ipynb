```python
numbers = [10, 15, 22, 33, 40, 51, 64, 75]
print("All numbers in the list:")
for n in numbers:
    print(n)

total = 0
for n in numbers:
    total = total + n

print("\nSum of all numbers:", total)

print("\nEven numbers are:")
for n in numbers:
    if n % 2 == 0:
        print(n)
```

```python
student = {
    "name": "Ahmed",
    "age": 20,
    "course": "Python",
    "marks": [65, 70, 80]
}

print("Student Details:")
for key in student:
    print(key, ":", student[key])

marks_list = student["marks"]
total = 0
for m in marks_list:
    total = total + m

average = total / len(marks_list)
print("\nAverage Marks:", average)

if average >= 50:
    print("Result: Passed")
else:
    print("Result: Failed")
```

```python
employees = [
    {"name": "Ali", "department": "IT", "salary": 50000},
    {"name": "Sara", "department": "HR", "salary": 65000},
    {"name": "Bilal", "department": "Accounts", "salary": 55000}
]

print("All Employees:")
for emp in employees:
    print(emp["name"], "-", emp["department"], "-", emp["salary"])

highest_salary = 0
highest_emp = ""

for emp in employees:
    if emp["salary"] > highest_salary:
        highest_salary = emp["salary"]
        highest_emp = emp["name"]

print("\nHighest Paid Employee:", highest_emp, "with salary", highest_salary)

total_salary = 0
for emp in employees:
    total_salary = total_salary + emp["salary"]

print("Total Salary Expense:", total_salary)
```

```python
numbers = []

print("Enter numbers (enter -1 to stop):")

while True:
    num = int(input("Enter number: "))
    if num == -1:
        break
    numbers.append(num)

print("\nComplete List:", numbers)

largest = numbers[0]
for n in numbers:
    if n > largest:
        largest = n

print("Largest Number:", largest)

smallest = numbers[0]
for n in numbers:
    if n < smallest:
        smallest = n

print("Smallest Number:", smallest)
```

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
word_count = {}

for word in words:
    if word in word_count:
        word_count[word] = word_count[word] + 1
    else:
        word_count[word] = 1

print("\nWord Frequency:")
for w in word_count:
    print(w, ":", word_count[w])
```
